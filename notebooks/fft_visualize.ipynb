# FFT Visualize - Amplitude / Phase debug

Notebook nay dung de:
1. Doc 1 anh thuc (Live face), tach Amplitude/Phase qua `FrequencyUtils.fft_decompose`.
2. Quan sat pho bien do (log-scale) va pho pha.
3. Thu ghep Amplitude_fake (gia lap) + Phase thuc -> xem ket qua IFFT co hop ly khong,
   truoc khi dua amplitude thuc su sinh ra tu Diffusion model vao.

```python
import sys
sys.path.append('../src')

import torch
import matplotlib.pyplot as plt

from improved_ddpm.fas.frequency_utils import FrequencyUtils
```

```python
# Dummy image de test nhanh (thay bang anh thuc tu data/raw/ khi can)
img = torch.rand(1, 3, 64, 64) * 2 - 1

amplitude, phase = FrequencyUtils.fft_decompose(img)
log_amp = FrequencyUtils.log_amplitude(amplitude)

print('amplitude shape:', amplitude.shape)
print('phase shape:', phase.shape)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(12, 4))

axes[0].imshow((img[0].permute(1, 2, 0) + 1) / 2)
axes[0].set_title('Input image')

axes[1].imshow(log_amp[0, 0], cmap='viridis')
axes[1].set_title('Log-Amplitude (channel R)')

axes[2].imshow(phase[0, 0], cmap='twilight')
axes[2].set_title('Phase (channel R)')

for ax in axes:
    ax.axis('off')
plt.tight_layout()
plt.show()
```

```python
# Kiem tra round-trip: fft_decompose -> ifft_reconstruct phai khoi phuc anh ban dau
recon = FrequencyUtils.ifft_reconstruct(amplitude, phase)
print('Max sai so tuyet doi:', (img - recon).abs().max().item())
```
